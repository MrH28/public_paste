## Exercício
Verifique se o kernel selecionado no VS Code corresponde ao ambiente virtual do projeto.


```python
from pathlib import Path
import os
import subprocess
import sys

# O interpretador que executa esta célula é o kernel atualmente selecionado.
python_executable = Path(sys.executable).resolve()
project_root = Path.cwd().resolve()
virtual_env = os.environ.get("VIRTUAL_ENV")

# Procura o ambiente virtual a partir da pasta de trabalho e das pastas-pai.
virtual_env_path = Path(virtual_env).resolve() if virtual_env else None
if virtual_env_path is None:
    for parent in (project_root, *project_root.parents):
        for candidate_name in (".venv", "venv", "env"):
            candidate = parent / candidate_name
            if candidate.exists():
                virtual_env_path = candidate.resolve()
                break
        if virtual_env_path is not None:
            break

if virtual_env_path is None:
    raise EnvironmentError(
        "Nenhum ambiente virtual foi encontrado. Ative o ambiente do projeto "
        "antes de iniciar o kernel do notebook."
    )

# Garante que o pacote necessário para o kernel esteja instalado neste Python.
subprocess.check_call([
    str(python_executable), "-m", "pip", "install", "--quiet", "ipykernel"
])

kernel_name = f"{project_root.name}-venv"
display_name = f"Python ({project_root.name} - ambiente virtual)"
subprocess.check_call([
    str(python_executable), "-m", "ipykernel", "install", "--user",
    "--name", kernel_name,
    "--display-name", display_name,
])

is_matching_kernel = python_executable.parent.parent == virtual_env_path
print(f"Python em execução: {python_executable}")
print(f"Ambiente virtual detectado: {virtual_env_path}")
print(f"Kernel registrado: {display_name}")
print(f"O kernel corresponde ao ambiente virtual: {'SIM' if is_matching_kernel else 'NÃO'}")

if not is_matching_kernel:
    print(
        "No VS Code, use 'Python: Select Interpreter' e escolha o Python deste ambiente; "
        "depois selecione o kernel com o mesmo nome no canto superior direito."
    )
```
